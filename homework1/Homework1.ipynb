## Q6. Median value of horsepower
Does the median change after filling missing values with the most frequent value?

```python
median_before = df["horsepower"].median()
mode = df["horsepower"].mode()[0]

df["horsepower"] = df["horsepower"].fillna(mode)

median_after = df["horsepower"].median()
median_before, mode, median_after
```

## Q7. Sum of weights
Linear regression with the normal equation: `w = (XᵀX)⁻¹ Xᵀ y`

```python
import numpy as np

X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).to_numpy()
XTX = X.T @ X
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = np.linalg.inv(XTX) @ X.T @ y
w.sum()
```
